# **Exploratory Data Analysis**

Exploratory Data Analysis (EDA) is a method used to `analyze` and `summarize` datasets. In python, you can use the Pandas library to perform EDA on a dataset. 

```python
# import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

### **1. Analysis of Data**

```python
# import dataset
df=sns.load_dataset('titanic')
df
```

### **3. Handling of Missing/Null Values**

#### **a. Find Missing/Null Values**

```python
# boolean
df.isnull()
```

```python
df.isnull().sum()
```

#### **b. Finding Percentage of Missing/Null Values**

```python
(df.isnull().sum()/len(df))*100
```

#### **c. Visualize the NaN values in the Heatmap**

```python
sns.heatmap(df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
# remove 'deck' column as it contains more than 70% NaN values
df.drop(columns='deck', inplace=True)
```

```python
df.columns
```

```python
# visualize and explore 
sns.countplot(data=df, x='survived')
```

```python
sns.barplot(data=df, x='survived', y='age', hue='who')
```

```python
# In bar plot, categorical data is provided at x axis and numeric data at y axis
sns.barplot(data=df, x='survived', y='fare', hue='who')
```

```python
# In count plot, only categorical data is provided
sns.countplot(x='pclass', data=df)
```

```python
# data distribution 
sns.histplot(data=df, x='age', bins=30)
```

```python
# Numeric data is required for histplot
df['age'].hist(bins=30, color= 'darkred', alpha=0.7)
```

```python
# boxplot display categorical data on x axis
# sns.boxplot(data=df, y='age', x='who')
# sns.boxplot(data=df, y='age', x='survived')
sns.boxplot(data=df, y='age', x='pclass', hue='who')
```

```python
# scatter plot required numeric data both at x axis and y axis
sns.scatterplot(data=df, x='age', y='fare', hue='pclass')
```

```python
# scatter plot required numeric data both at x axis and y axis
sns.scatterplot(data=df, x='age', y='fare', hue='pclass')
plt.ylim(0,300)
```

```python
sns.pairplot(df, hue='survived')
```

```python
# stacked barchart/plot
# how to change position in x-stick in matplotlib
df.groupby(['pclass','who'])['fare'].mean(). unstack().plot(kind='bar', stacked=True)
```

```python
# swarm plot
sns.swarmplot(data=df, x='survived', y='age', hue='who')
```

```python
# point plot
sns.catplot(data=df, x='pclass', y='fare', hue='embarked', kind='point')
```

```python
# relation of two numerical variables
sns.lmplot(data=df, y='fare', x='age')
```

```python
# conditional argument in sns.lmplot 
sns.lmplot(data=df[df['fare'] < 300], y='fare', x='age', hue='pclass')
```

```python
# grouping
# how to add a column and make a age group
df.columns
```

```python
df['age_groups']= pd.cut(df['age'], [0,10,18,30,50,80],labels=['Kids', 'Child', 'Youth', 'Middle aged', 'Old'])
```

```python
df.head()
```

```python
df.columns
```

```python
df.groupby('age_groups')['survived'].mean().plot(kind='bar')
```

```python
sns.pairplot(data=df, hue='survived', vars=['age', 'fare', 'pclass'])
```
