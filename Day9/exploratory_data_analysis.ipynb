#### **4. Handling Missing or Null Values**

A missing value is an absence of data in a dataset. In other words, it’s a place where information should exist but doesn’t.

##### **Reasons for Missing Values**

1. Data collection issues: Sensor malfunction, network failure.
2. Data entry errors: Typos, misinterpretation during manual entry.
3. Privacy concerns: Respondents skip sensitive questions (e.g., income).
4. Survey design flaws: Optional fields left blank.
5. Data integration: Merging multiple sources with inconsistent information.
6. System limitations: API failure, corrupted files.
7. Intentional omission: When people deliberately hide information.

##### **How Do Missing Values Impact Analysis?**

1. Training difficulties: Many ML algorithms can’t handle missing data.
2. Increased preprocessing: Extra effort required to fix incomplete records.
3. Bias in results: Incorrect imputations may lead to false insights.
4. Misleading outcomes: Wrong predictions due to improper handling.
5. Loss of information: Dropping too many rows/columns reduces data size.

##### **c. General Rules for Handling Missing Values**
1. Above 70–75%: Consider dropping the column.
2. Around 2%: Consider dropping rows using dropna().
3. Below 30–40%: Consider imputing using mean, median, mode, or other suitable methods.
  
- Mean: Suitable for approximately normal numerical data.
- Median: Suitable for skewed numerical data and data with outliers.
- Mode: Suitable for categorical data.

5. Between 40–70%: Evaluate carefully; consider imputation or dropping the column.
   
Remember: These are general guidelines, not fixed rules. Always consider the importance of the data and the reason for missing values.
