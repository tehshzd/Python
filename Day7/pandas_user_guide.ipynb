### **3. pd.date_range()**

It is a Pandas function used to generate a sequence of dates at a specified frequency

#### **Syntax**

pd.date_range(start, periods, freq)

1. start: starting date

2. periode: generates demanded date/time entries

3. freq: determines the interval between the entries


| Frequency | Meaning |
|-----------|---------|
| `D`       | Daily |
| `2D`      | Every 2 days |
| `3D`      | Every 3 days |
| `W`       | Weekly |
| `MS`      | Month Start |
| `ME`      | Month End |
| `QS`      | Quarter Start |
| `QE`      | Quarter End |
| `YS`      | Year Start |
| `YE`      | Year End |
| `h`       | Hourly |
| `min`     | Every minute |

```python
dates=pd.date_range("20130101", periods=6, freq='ME')
dates
```

### **4. Creating a DataFrame with Random Numbers**

#### **Syntax**

1. **pd.DataFrame()**: Creates a two-dimensional DataFrame



2. **np.random.randn(6, 4)**: Generates 6 rows and 4 columns of random numbers from a standard normal distribution



3. **index=dates**: Assigns dates as row indexes



4. **columns=list("ABCD")**: Assigns A, B, C, and D as column names

```python
df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list("ABCD"))
df
```

#### **5. Creating a DataFrame Using a Dictionary**

A Pandas DataFrame can be created using a dictionary containing different data types

- `pd.Timestamp()`: Converts a date string into a Timestamp object
  
- `pd.Series()`: Creates a one-dimensional labelled array
  
- `range(4)`: Generates numbers from 0 to 3
  
- `list()`: Converts an iterable into a list
  
- `dtype="float32"`: Specifies 32-bit floating-point data
  
- `np.array()`: Creates a NumPy array
  
- `[3] * 4`: Repeats 3 four times
  
- `dtype="int32"`: Specifies 32-bit integer data
  
- `pd.Categorical()`: Creates categorical data
  
- `"foo"`: Represents a string value


```python
df2 = pd.DataFrame(
    {
        "A": 1.0,
        "B": pd.Timestamp("20130102"),
        "C": pd.Series(1, index=list(range(4)), dtype="float32"),
        "D": np.array([3] * 4, dtype="int32"),
        "E": pd.Categorical(["test", "train", "test", "train"]),
        "F": "foo",
    }
)


df2
```

### **Difference between `iloc` and `loc`**

| `iloc` | `loc` |
|---|---|
| Integer Location | Label-based Location |
| Selects by integer position | Selects by index/column labels |
| Uses numerical positions | Uses actual labels |
| `df.iloc[0:3]` | `df.loc[0:3]` |
| Stop position is excluded | Stop label is generally included in slicing |

#### Column Selection Using `iloc`

| Syntax | Meaning |
|---|---|
| `df.iloc[:, 0:3]` | Selects columns at positions 0, 1, 2 |
| `df.iloc[:, :3]` | Selects the first 3 columns |
| `df.iloc[:, 0:]` | Selects all columns starting from position 0 |
| `df.iloc[:, :]` | Selects all rows and all columns |
| `df.iloc[:, 1:4]` | Selects columns at positions 1, 2, 3 |
| `df.iloc[:, 2:]` | Selects all columns starting from position 2 |
| `df.iloc[:, :2]` | Selects the first 2 columns |

#### Row Selection Using `iloc`

| Syntax | Meaning |
|---|---|
| `df.iloc[0:3]` | Selects rows at positions 0, 1, 2 |
| `df.iloc[:3]` | Selects the first 3 rows (0, 1, 2) |
| `df.iloc[0:]` | Selects all rows starting from position 0 |
| `df.iloc[:]` | Selects all rows |
| `df.iloc[:, 1:4]` | Selects all rows and columns at positions 1, 2, 3 |

```python
# selecting required rows and columns from dataframe
# first row indexed as 0, first column indexed as 0
# second row indexed as 1, second column indexed as 1 and so on
# iloc is index loc function
# In Pandas iloc, the stop index is excluded (not included)
# 3rd row and 4rth column are stop indexes

df.iloc[1:3, 1:4]
```

```python
# ':' means all rows in the dataframe 
# '0: means all rows in the dataframe
# df.iloc[:, 0:3] 
df.iloc[0:, 1:4]
```

```python
df.iloc[3:5, 0:2]
```
