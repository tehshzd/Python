### Missing Value (nan)/ Null Value

nan stands for Not a Number. In Pandas, it is commonly used to represent a missing or unavailable value in a 
dataset.

The importance of handling missing (NaN) data in a dataset: 

1. Accuracy: Missing values can affect the accuracy of machine learning models.
   
2. Data quality: Handling NaN values makes the dataset more complete.

3. Model performance: Some ML algorithms cannot work with missing values.

4. Avoid bias: Improper handling of missing data can lead to misleading results.
