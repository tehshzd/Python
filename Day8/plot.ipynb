```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import dataset 
titanic=sns.load_dataset('titanic')
titanic
```

```python
titanic.info()
```

```python
# missing/null values
titanic.isnull().sum()
```

```python
# bar plot 
titanic.survived.value_counts().plot(kind='bar')
```

```python
# bar plot
sns.countplot(x='survived', hue='sex', data=titanic)

```

```python
# scatter plot: x and y variables must be integer or float
plt.scatter(data=titanic, x='age', y='fare')
```

```python
sns.scatterplot(data=titanic, x='age', y='fare', hue='pclass')
```

```python
sns.scatterplot(data=titanic, x='sex', y='fare', hue='pclass')
```

```python
sns.barplot(data=titanic, x='sex', y='fare', hue='pclass')
```

```python
# pair plot
sns.pairplot(titanic)
```

```python
# histogram
# titanic.hist()
# titanic.hist(bins=10)
# titanic.hist(bins=10, figsize=(9,7))
titanic.hist(bins=10, figsize=(9,7), grid=False)
```

```python
# box plot
sns.boxplot(data=titanic, x='sex', y='fare', hue='pclass')
```

```python
sns.boxplot(data=titanic, x='sex', y='fare', hue='parch')
```

```python
sns.boxplot(data=titanic, x='sex', y='fare', hue='who')
```

```python
# boxen plot
sns.boxplot(data=titanic, x='sex', y='fare', hue='class')
```

```python
# Grid Plot
g = sns.FacetGrid (titanic, hue="survived", col="pclass", margin_titles=True)
pallete = {1:"yellow", 0:"red"}

g = g.map(plt.scatter, "fare", "age", edgecolor="w"). add_legend();
```

```python
g = sns.FacetGrid (titanic, hue="survived", col="sex", margin_titles=True)
pallete = {1:"yellow", 0:"red"}

g = g.map(plt.scatter, "fare", "age", edgecolor="w"). add_legend();
```

```python
g = sns.FacetGrid (titanic, hue="survived", col="who", margin_titles=True)
pallete = {1:"yellow", 0:"red"}

g = g.map(plt.scatter, "fare", "age", edgecolor="w"). add_legend();
```

```python
# strip plot
sns.stripplot(data=titanic,x='sex',y='fare', hue='class')
```

```python
# heat map
corr=titanic.corr(numeric_only=True)
plt.figure(figsize=(10,10))
sns.heatmap(corr)
```

```python
corr=titanic.corr(numeric_only=True)
plt.figure(figsize=(10,10))
sns.heatmap(corr)
plt.title('Heat Map of Titanic')
```

```python
corr = titanic.corr(numeric_only=True)
plt.figure(figsize=(10, 10))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.violinplot(data=titanic, x='sex', y='fare', hue='pclass')
```
