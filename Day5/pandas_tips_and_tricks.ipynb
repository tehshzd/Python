# 16. Splitting a String into Multiple Columns

```python
# import libraries
import pandas as pd

df=pd.DataFrame({'name': ['Tehreem Shehzad', 'Tyba Akbar', 'Rabia Akhlak'], 'location': ['Shorkot, Pakistan', 'Jhang, Pakistan', 'Lahore, Pakistan']})
df 
```

```python
df.name.str.split(' ').head()
```

```python
df.name.str.split(' ').expand=True
df
```

```python
# splitting a column into multiple columns using the str.split() method with the expand parameter set to True. 
# This will create a new DataFrame with separate columns for each part of the split string.
df.name.str.split(' ', expand=True)
```

```python
# adding those split columns back to the original DataFrame as new columns named 'first_name' and 'last_name'.
df[['first_name', 'last_name']] = df.name.str.split(' ', expand=True)
df
```

```python
df.location.str.split(' ', expand=True)
```

```python
df[['city', 'country']] = df.location.str.split(', ', expand=True)
df
```

```python
# Refine data manipulation by splitting the 'location' column into separate 'city' and 'country' columns, 
# and 'name' column into 'first_name' and 'last_name', allowing for easier analysis and filtertion.
df[['first_name', 'last_name', 'city', 'country']]
```
